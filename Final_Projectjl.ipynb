## Objective
Decode an English-language message that has been encrypted using substitution cipher, using character frequency analysis, patter recognition, and iterative mapping.

## Approach
1. Count character frequencies in the encoded message
2. Make initial guesses based on common English words (the, and, this)
3. Apply the mapping and examine partially decoded text
4. Iteratively refine the mapping using context clues
5. Produce the final decoded message


```python
from collections import Counter

encoded_msg = "jyn fg jggtwj djtfcn stf sjyn edcyjnc ia zy stes fjqtye z wzdn owcff gstf gsq sjyn edcyjnc gsjg mtgs tg gszi xjqcfg owzm gstyc cycxtcf gz gtyq otgf ty gsq xcdhq jyn gsc wzdn ntn edty jyn aczawc ntn lcjfg iazy gsc wjxof jyn fwzgsf jyn hjda jyn jyhszktcf jyn zdjyeigjyf jyn odcjvljfg hcdcjwf jyn lditg ojgf jyn gsc wzdn fajvc fjqtye ltdfg fsjwg gszi gjvc zig gsc szwq aty gscy fsjwg gszi hziyg gz gsdcc yz xzdc yz wcff gsdcc fsjwg oc gsc yixocd gszi fsjwg hziyg jyn gsc yixocd zl gsc hziygtye fsjwg oc gsdcc lzid fsjwg gszi yzg hziyg yctgscd hziyg gszi gmz cphcagtye gsjg gszi gscy adzhccn gz gsdcc ltkc tf dtesg zig zyhc gsc yixocd gsdcc octye gsc gstdn yixocd oc dcjhscn gscy wzoocfg gszi gsq szwq sjyn edcyjnc zl jygtzhs gzmjdnf gszi lzc msz octye yjiesgq ty xq ftesg fsjww fyill tg"

# Letter frequency
letters = [c for c in encoded_msg if c.isalpha()]
freq = Counter(letters)

print("Total letters:", len(letters))
print("\nLetter frequencies (most common first):")
for letters, count in freq.most_common():
  print(f" {letters}: {count}")

# Word frequency
words = encoded_msg.split()
word_freq = Counter(words)

print("\nMost common words:")
for word, count in word_freq.most_common(15):
  print(f" {word}: {count}")
```

## Initial Observations

The most frequent 3-letter words are 'gsc' and 'jyn'. In English, the most common 3-letter words are **the** and **and**.

- 'gsc' = the -> g:t, s:h, c:e
- 'jyn' = and -> j:a, y:n, n:d

other frequent words:
- 'gyzi' (4 letters, starts with gs) -> likely "thou"
- 'gstf' (4 letters, starts with gs) -> likely "this"
- 'syjn' (4 letters) with s:h, j:a, y:n, n:d = "hand"


```python
# Initial mapping
mapping = {
    'g':'t','s':'h','c':'e', # gsc = the
    'j':'a','y':'n','n':'d', # jyn = and
    'z':'o','i':'u',    # gszi = thou
    't':'i','f':'s', # gstf = this
    'd':'r',  # gsdcc = three
}

def decode(msg, mapping):
  result = []
  for ch in msg:
    if ch in mapping:
      result.append(mapping[ch])
    elif ch == ' ':
      result.append(' ')
    else:
      result.append('?') # unknown
  return ''.join(result)

print(decode(encoded_msg, mapping))
```

```python
# Complete mapping (encoded -> plaintext)
mapping = {
    'a':'p',
    'c':'e',
    'd':'r',
    'e':'g',
    'f':'s',
    'g':'t',
    'h':'c',
    'i':'u',
    'j':'a',
    'k':'v',
    'l':'f',
    'm':'w',
    'n':'d',
    'o':'b',
    'q':'y',
    's':'h',
    't':'i',
    'v':'k',
    'w':'l',
    'x':'m',
    'y':'n',
    'z':'o',
}

def decode(msg, mapping):
  return ''.join(mapping.get(ch, ch) if ch != ' ' else ' ' for ch in msg)

decoded_message = decode(encoded_msg, mapping)
print(decoded_message)
```
